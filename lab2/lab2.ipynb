## Task4. Filtering

#### Question 1: How many passengers were adults?

```python
(df[df['Age'] >= 18].count()).iloc[0]
```

#### Question 2: How many passengers were embarked in Queenstown?

```python
df[df['Embarked'] == 'Q'].count().iloc[0]
```

#### Question 3: How many woman passengers survived? 

```python
df[(df['Sex'] == 'female') & (df['Survived'] == 1)].count().iloc[0]
```

#### Question 4: Which passengers had fare higher than the median?

```python
print(f"Median: {df['Fare'].median()}")
df[df['Fare'] > df['Fare'].median()]
```

## Task 5. EDA

#### Missing values and duplicates

```python
df_clear = df
```

```python
df.info()
```

```python
df_clear['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df_clear['Age'].isna().sum()
```

```python
df_clear = df_clear.drop(columns=['Cabin'])
```

```python
df_clear.info()
```

```python
df_clear = df_clear[df_clear['Embarked'].notna()]
```

```python
df_clear.info()
```

```python
df_clear.duplicated().sum()
```

#### Numerical values statistics

```python
df_clear[['Age', 'SibSp', 'Parch', 'Fare']].describe()
```

#### Categorical values statistics

```python
categories = ['Survived', 'Pclass', 'Sex', 'Embarked']

for category in categories:
    df_clear[category] = df_clear[category].astype('category')
```

```python
df_clear[categories].describe()
```

#### Relationships

```python
df_clear[['Age', 'Fare', 'Pclass', 'Survived']].corr()
```

```python
df_clear.groupby('Sex')['Survived'].value_counts()
```
